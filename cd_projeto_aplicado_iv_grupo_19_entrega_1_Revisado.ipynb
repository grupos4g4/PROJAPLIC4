# **Introdução**

A educação superior brasileira atravessa uma década de transformações marcadas pela expansão da modalidade a distância (EAD), recomposição de matrículas após os choques da pandemia e ajustes na rede de instituições e cursos. O Censo da Educação Superior, conduzido anualmente pelo Inep, é a principal fonte para compreender essas dinâmicas e subsidiar a formulação, o monitoramento e a avaliação de políticas públicas, além de apoiar estudos acadêmicos e setoriais (BRASIL, 2024).

No horizonte 2013–2023, destacam-se tendências de longo prazo em vagas/ingressantes, matrículas e concluintes, organizadas em séries históricas consolidadas nas Notas Estatísticas e nos microdados oficiais, com recortes por modalidade (presencial/EAD), categoria administrativa e organização acadêmica (BRASIL, 2024).

# **Motivação (Com base setorial e estatísticas)**

Análises recentes com base nos números do Censo indicam que, após a contração observada em 2020, houve recuperação do setor e manutenção do número de IES credenciadas entre 2021 e 2023; ainda assim, persistem assimetrias entre redes pública e privada e entre modalidades (CHANAN, 2023). Complementarmente, o Censo EAD.BR 2023/2024 documenta práticas de gestão, docência e tecnologia associadas a cursos EAD com desempenho máximo no Enade — como organização de equipes multidisciplinares, arranjos de tutoria e uso de tecnologias — oferecendo hipóteses explicativas e referências de boas práticas para orientar intervenções (ABED, 2024).

# **Justificativa (Contribuições Técnicas e Sociais)**

O projeto realiza análise de séries temporais com decomposição (tendência/sazonalidade/ciclos/ruído), detecção de quebras e previsão de curto prazo, combinando objetivos de modelagem e previsão, conforme os conceitos metodológicos trabalhados no componente (UNIVERSIDADE, 2025).

*Contribuição Social*. Ao transformar séries do Censo em diagnósticos e cenários prospectivos, o projeto fornece evidências para decisões sobre acesso, permanência e conclusão, apoiando redução de evasão e melhor uso de infraestrutura (presencial e digital), em alinhamento com os ODS trabalhados no componente (UNIVERSIDADE, 2025).


# **Objetivo**

*Geral.* Modelar e analisar as séries 2013–2023 de ingressantes, matrículas e concluintes no ensino superior brasileiro, com recortes por modalidade, categoria administrativa e organização acadêmica, para identificar tendências, quebras e diferenças estruturais, além de prever cenários de curto prazo (BRASIL, 2024).

*Específicos*.
